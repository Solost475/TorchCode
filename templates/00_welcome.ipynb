# 🔥 TorchCode — PyTorch 算子练习

欢迎使用 TorchCode！这是一个基于 JupyterLab 的 PyTorch 算子实现练习环境。

## 📁 文件结构

每道题都有**两个**笔记本：

| 文件 | 说明 |
|------|-------------|
| `01_relu.ipynb` | ✏️ **空白模板**：在这里编写你的实现 |
| `01_relu_solution.ipynb` | 📖 **参考解答**：遇到困难时可查阅 |

## 使用方法

1. **打开题目**：选择一个空白模板，如 `01_relu.ipynb`。
2. **阅读题面**：第一个单元格包含题目描述和函数或类接口。
3. **编写代码**：在 `✏️ 在此编写你的实现` 单元格中完成实现。
4. **调试**：可自由插入单元格，使用 `print(x.shape)` 等语句检查结果。
5. **提交**：运行最后一个单元格中的 `check("relu")`，查看判题结果。
6. **遇到困难**：调用 `hint("relu")` 查看提示，或打开以 `_solution.ipynb` 结尾的参考解答。
7. **重新练习**：点击工具栏中的 🔄 **Reset（重置）** 按钮，恢复空白模板后再次练习。

> 💡 每个笔记本都有 **Colab** 工具栏按钮和**在 Colab 中打开**徽章，可直接在 Google Colab 中练习，无需本地配置。
