### 模块结构（预归一化）
```
x = x + causal_self_attention(ln1(x))
x = x + mlp(ln2(x))
```

### 函数或类接口
```python
class GPT2Block(nn.Module):
    def __init__(self, d_model: int, num_heads: int): ...
    def forward(self, x: torch.Tensor) -> torch.Tensor: ...
```

### 实现要求
- 必须继承 `nn.Module`。
- `self.ln1`, `self.ln2`: `nn.LayerNorm(d_model)`
- `self.W_q`、`self.W_k`、`self.W_v` 和 `self.W_o`：使用 `nn.Linear` 实现注意力投影。
- `self.mlp`: `nn.Sequential(Linear(d, 4d), GELU(), Linear(4d, d))`
- 注意力必须满足**因果约束**，屏蔽未来位置。
- 使用预归一化结构，在注意力和 MLP **之前**应用 LayerNorm。
- 注意力和 MLP 都需要残差连接。
