# 🟡 中等：SwiGLU 前馈网络

实现 LLaMA 等现代大语言模型使用的 **SwiGLU MLP（前馈网络）**。

$$
\text{SwiGLU}(x) = \text{down\_proj}\big(\text{SiLU}(\text{gate\_proj}(x)) \odot \text{up\_proj}(x)\big)
$$

其中，$\text{SiLU}(x) = x \cdot \sigma(x)$。

### 函数或类接口
```python
class SwiGLUMLP(nn.Module):
    def __init__(self, d_model: int, d_ff: int): ...
    def forward(self, x: torch.Tensor) -> torch.Tensor: ...
```

### 实现要求
- 必须继承 `nn.Module`。
- `self.gate_proj`: `nn.Linear(d_model, d_ff)`
- `self.up_proj`: `nn.Linear(d_model, d_ff)`
- `self.down_proj`: `nn.Linear(d_ff, d_model)`
- 激活函数使用 **SiLU**（也称 Swish）：可调用 `F.silu`，或实现为 `x * torch.sigmoid(x)`。

### 为什么使用 SwiGLU？
与传统的 `Linear → ReLU/GELU → Linear` 前馈网络相比，SwiGLU 使用**门控机制**：
门控投影控制信息流，升维投影提供具体内容。
在实践中，这种结构通常优于标准前馈网络，PaLM、LLaMA 和 Mistral 都使用了它。
