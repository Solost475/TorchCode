### 函数或类接口
```python
class SimpleLinear:
    def __init__(self, in_features: int, out_features: int): ...
    def forward(self, x: torch.Tensor) -> torch.Tensor: ...
```

### 实现要求
- `self.weight`：形状为 `(out_features, in_features)`，使用 `randn * (1/√in_features)` 初始化。
- `self.bias`：形状为 `(out_features,)`，初始化为零。
- 两者都必须设置 `requires_grad=True`。
- `forward(x)` 计算 `x @ W^T + b`。
- **禁止**使用 `torch.nn.Linear`。
