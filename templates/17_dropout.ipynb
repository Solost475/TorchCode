# 🟢 简单：随机失活（Dropout）

从零实现 **Dropout（随机失活）** 正则化。

### 函数或类接口
```python
class MyDropout(nn.Module):
    def __init__(self, p: float = 0.5): ...
    def forward(self, x: Tensor) -> Tensor: ...
```

### 约束条件
- **训练时**：以概率 `p` 将每个元素置零，其余元素乘以 `1/(1-p)`。
- **推理时**：原样返回输入，即恒等映射。
- **禁止**使用 `nn.Dropout` 或 `F.dropout`。
