# 🔴 困难：分组查询注意力（GQA）

实现**分组查询注意力（GQA）**，LLaMA 2、Mistral 等模型用它来减小 KV 缓存。

它与 MHA 类似，但 **KV 头的数量少于 Q 头**。每组 Q 头共享同一个 K/V 头。

### 函数或类接口
```python
class GroupQueryAttention:
    def __init__(self, d_model: int, num_heads: int, num_kv_heads: int): ...
    def forward(self, x) -> torch.Tensor:  # 自注意力
```

### 实现要求
- `self.W_q`：`nn.Linear(d_model, d_model)`，将 Q 投影到完整维度。
- `self.W_k`：`nn.Linear(d_model, num_kv_heads * d_k)`，将 K 投影到较小维度。
- `self.W_v`：`nn.Linear(d_model, num_kv_heads * d_k)`，将 V 投影到较小维度。
- `self.W_o`：`nn.Linear(d_model, d_model)`，用于输出投影。
- `d_k = d_model // num_heads`
- 使用 `repeat_interleave` 扩展 KV 头，使其数量与 Q 头匹配。
- 当 `num_kv_heads == num_heads` 时，行为应与标准 MHA 一致。
