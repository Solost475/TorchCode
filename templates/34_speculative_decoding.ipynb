### 算法步骤
对每个位置 i = 0, ..., K-1：
1. `ratio = target_probs[i, token_i] / draft_probs[i, token_i]`

3. 若拒绝，则从 `normalize(max(0, target - draft))` 中采样，将结果追加到列表后停止。
