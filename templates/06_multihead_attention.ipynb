# 🔴 困难：多头注意力（MHA）

从零实现**多头注意力（Multi-Head Attention）**，它是 Transformer 的核心组成部分。

$$
\text{MultiHead}(Q, K, V) = \text{Concat}(\text{head}_1, \dots, \text{head}_h) W^O
$$

$$
\text{head}_i = \text{Attention}(Q W_i^Q,\; K W_i^K,\; V W_i^V)
$$

### 函数或类接口
```python
class MultiHeadAttention:
    def __init__(self, d_model: int, num_heads: int): ...
    def forward(self, Q, K, V) -> torch.Tensor: ...
```

### 实现要求
- 使用 `nn.Linear(d_model, d_model)` 创建 `self.W_q`、`self.W_k`、`self.W_v` 和 `self.W_o`。
- 每个头的维度为 `d_k = d_model // num_heads`。
- `forward(Q, K, V)`：Q 的形状为 `(B, seq_q, d_model)`，K/V 的形状为 `(B, seq_k, d_model)`。
- 必须支持**交叉注意力**（`seq_q != seq_k`）。
- **禁止**使用 `torch.nn.MultiheadAttention`。
- **允许**使用 `torch.softmax` 和 `torch.matmul`。

### 实现步骤
1. 投影：`q = self.W_q(Q)`、`k = self.W_k(K)`、`v = self.W_v(V)`。
2. 重塑为 `(B, num_heads, seq, d_k)`。
3. 对每个头计算缩放点积注意力。
4. 拼接各头，得到 `(B, seq_q, d_model)`。
5. 输出投影：`self.W_o(concat)`。
