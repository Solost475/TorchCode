# 🟢 参考解答：ReLU 激活函数

ReLU 激活函数的参考解答。

$$
\text{ReLU}(x) = \max(0, x)
$$

```python
import torch
```

```python
# ✅ 参考解答

def relu(x: torch.Tensor) -> torch.Tensor:
    return x * (x > 0).float()
```

```python
# 验证
x = torch.tensor([-2., -1., 0., 1., 2.])
print("Input: ", x)
print("Output:", relu(x))
```
